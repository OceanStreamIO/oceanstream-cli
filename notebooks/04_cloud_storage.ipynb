## STAC Metadata with Cloud Storage

When uploading to cloud storage, STAC asset hrefs are automatically converted to relative paths:

```json
{
  "assets": {
    "pmtiles": {
      "href": "../tiles/track.pmtiles",
      "type": "application/vnd.pmtiles"
    },
    "data": {
      "href": "../lat_bin=45/lon_bin=-122/part-0.parquet",
      "type": "application/x-parquet"
    }
  }
}
```

This ensures portability - the STAC catalog works regardless of where it's hosted.
